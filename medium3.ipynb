```python

import pandas as pd


df = pd.read_json("imdb_movies_2000to2022.prolific.json", lines=True)


df = df[df['title'].notna()]
df = df[df['genres'].map(lambda x: isinstance(x, list) and len(x) > 0)].reset_index(drop=True)


df[['title', 'genres']].head()

def jaccard_similarity(set1, set2):
    intersection = len(set(set1).intersection(set(set2)))
    union = len(set(set1).union(set(set2)))
    return intersection / union if union > 0 else 0

def find_similar_movies(df, query_title, top_n=10):
    if query_title not in df['title'].values:
        return f"'{query_title}' not found in dataset."

    query_genres = df[df['title'] == query_title].iloc[0]['genres']
    similarities = []

    for idx, row in df.iterrows():
        if row['title'] != query_title:
            sim = jaccard_similarity(query_genres, row['genres'])
            similarities.append((row['title'], sim))

    top_similar = sorted(similarities, key=lambda x: x[1], reverse=True)[:top_n]
    return pd.DataFrame(top_similar, columns=["Similar Movie", "Jaccard Score"])

query_movies = ["Shrek", "Training Day", "Resident Evil"]

for movie in query_movies:
    print(f"\nTop 10 movies similar to '{movie}':")
    display(find_similar_movies(df, movie))
```
